## Test Weights
Goal: Test the weights and distributions

```python
# --- Standard Libraries ---
import os
import re
from collections import Counter

# --- Data Science ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

# --- Utils ---
from tqdm.auto import tqdm
from IPython.display import display

# --- Setup ---
tqdm.pandas()


final_daily_df = pd.read_csv(
    os.path.join("processed", "final_daily_df.csv"),
    parse_dates=["date"]
)

weighted_final_daily_df = pd.read_csv(
    os.path.join("processed", "weighted_final_daily_df.csv"),
    parse_dates=["date"]
)

print(final_daily_df.info())
print(weighted_final_daily_df.info())
```

## Create new Dataset with differences
Difference between weighted and unweigted

```python
# Geichtete cols fokussieren
weighted_cols = [
    "neg",
    "neu",
    "pos",
    "polarized",
    "anger",
    "disgust",
    "fear",
    "joy",
    "neutral",
    "sadness",
    "surprise",
    "Extroversion",
    "Neuroticism",
    "Agreeableness",
    "Conscientiousness",
    "Openness",
    "arts_culture",
    "business_entrepreneurs",
    "celebrity_pop_culture",
    "diaries_daily_life",
    "family",
    "fashion_style",
    "film_tv_video",
    "fitness_&_health",
    "food_&_dining",
    "gaming",
    "learning_educational",
    "music",
    "news_social_concern",
    "other_hobbies",
    "relationships",
    "science_technology",
    "sports",
    "travel_adventure",
    "youth_student_life",
]

# Nach „date“ indexieren
fd = final_daily_df.set_index("date")
wfd = weighted_final_daily_df.set_index("date")

# Differenz berechnen (final minus weighted)
diff_df = wfd[weighted_cols].subtract(fd[weighted_cols]).fillna(0)

# Optional: Index wieder als Spalte
diff_df = diff_df.reset_index()

# Spalten umbenennen, z.B. mit Suffix "_diff"
diff_df = diff_df.rename(
    columns={col: f"{col}_diff" for col in weighted_cols}
)

# Ergebnis anschauen
print(diff_df.info())

# Deine weighted_cols mit "_diff" erweitern
diff_cols = [f"{col}_diff" for col in weighted_cols]

# Plotte die korrekten Spalten
fig7 = px.line(
    diff_df,
    x="date",
    y=diff_cols,
    title="Tägliche Weighted Feature-Differenzen"
)
fig7.update_layout(xaxis_title="Datum", yaxis_title="Differenz")
fig7.show()
```

### Metrics
Check if the difference in each category balance each other out:
- Sentiment
- Emotions
- Personality
- Topics

```python

# 1) Liste der reinen Feature-Spalten (ohne "date")
feature_cols = diff_df.columns.drop("date")
sentiment_cols = [
    "neg_diff",
    "neu_diff",
    "pos_diff",
]
emotion_cols = [
    "anger_diff",
    "disgust_diff",
    "fear_diff",
    "joy_diff",
    "neutral_diff",
    "sadness_diff",
    "surprise_diff",
]
personality_cols = [
    "Extroversion_diff",
    "Neuroticism_diff",
    "Agreeableness_diff",
    "Conscientiousness_diff",
    "Openness_diff",
]
topics_cols = [
    "arts_culture_diff",
    "business_entrepreneurs_diff",
    "celebrity_pop_culture_diff",
    "diaries_daily_life_diff",
    "family_diff",
    "fashion_style_diff",
    "film_tv_video_diff",
    "fitness_&_health_diff",
    "food_&_dining_diff",
    "gaming_diff",
    "learning_educational_diff",
    "music_diff",
    "news_social_concern_diff",
    "other_hobbies_diff",
    "relationships_diff",
    "science_technology_diff",
    "sports_diff",
    "travel_adventure_diff",
    "youth_student_life_diff"
]

# 2) Summe aller Diffs pro Tag pro Kategorie
diff_df["sentiment_sum_diff"]   = diff_df[sentiment_cols].sum(axis=1)
diff_df["emotion_sum_diff"]     = diff_df[emotion_cols].sum(axis=1)
diff_df["personality_sum_diff"] = diff_df[personality_cols].sum(axis=1)
diff_df["topics_sum_diff"]      = diff_df[topics_cols].sum(axis=1)

# Die vier Spalten, die du über die Zeit plotten möchtest
plot_cols = [
    "sentiment_sum_diff",
    "emotion_sum_diff",
    "personality_sum_diff",
    "topics_sum_diff"
]

fig = px.line(
    diff_df,
    x="date",
    y=plot_cols,
    title="Tägliche Summen der Feature-Differenzen nach Kategorie"
)
fig.update_layout(
    xaxis_title="Datum",
    yaxis_title="Summe der Diffs",
    legend_title="Kategorie"
)
fig.show()
```

```python
cumsum_df = (
    diff_df
    .set_index("date")[feature_cols]
    .cumsum()               # akkumuliert über die Zeilen (Tage)
    .reset_index()          # macht date wieder zur Spalte
)

# Plotte die kumulierten Summen mit Plotly Express
fig_cumsum = px.line(
    cumsum_df,
    x="date",
    y=feature_cols,
    title="Kumulierte Summen der Feature-Differenzen über die Zeit"
)
fig_cumsum.update_layout(
    xaxis_title="Datum",
    yaxis_title="Kumulierte Summe der Diffs",
    legend_title="Feature"
)
fig_cumsum.show()
```
