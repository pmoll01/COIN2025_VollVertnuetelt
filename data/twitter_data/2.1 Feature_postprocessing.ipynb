```python
# personality values "normalisieren" (summe der personality values auf 1)
# kann ruhig zweimal ausgeführt werden. falls schon normalisiert, passiert nichts

import pandas as pd

# Big Five Spalten definieren
big5_cols = ["Openness", "Conscientiousness", "Extroversion", "Agreeableness", "Neuroticism"]

def normalize_big5(df, df_name=""):
    row_sums = df[big5_cols].sum(axis=1)

    # Prüfen, ob bereits normalisiert
    if (row_sums <= 1.01).all():
        print(f"{df_name}: Bereits normalisiert – keine Änderungen vorgenommen.")
    else:
        max_sum = row_sums.max()
        print(f"{df_name}: Maximale Zeilensumme = {max_sum:.4f} → wird normalisiert.")
        df[big5_cols] = df[big5_cols] / max_sum
    return df

# CSVs laden
weighted_final_daily_df = pd.read_csv("processed/weighted_final_daily_df.csv")
final_daily_df          = pd.read_csv("processed/final_daily_df.csv")

# Normalisierung durchführen
weighted_final_daily_df = normalize_big5(weighted_final_daily_df, "weighted_final_daily_df")
final_daily_df          = normalize_big5(final_daily_df, "final_daily_df")

# Optional speichern
weighted_final_daily_df.to_csv("processed/weighted_final_daily_df.csv", index=False)
final_daily_df.to_csv("processed/final_daily_df.csv", index=False)
```
