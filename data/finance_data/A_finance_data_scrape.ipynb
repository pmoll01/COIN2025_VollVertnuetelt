# Get all necesarry Finance Data and combine on one Frame

### get Data from Yahoo Finance Data

```python
import yfinance as yf

# S&P 500 Index from Yahoo Finance
sp500_data = yf.Ticker("^GSPC")

# Download historical data for the last 15 years
sp500_data = sp500_data.history(period="15y", interval="1d")

#print the first few rows of the data
print(sp500_data.head())
```

```python
import matplotlib.pyplot as plt

# Plot the 'Close' prices over time
plt.figure(figsize=(12, 6))
plt.plot(sp500_data.index, sp500_data['Close'], label='S&P 500 Close Price', color='blue')
plt.title('S&P 500 Close Price Over Time')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.legend()
plt.grid(True)
plt.show()
```

### combine in one Dataset

```python
import pandas as pd

# Normalize and align all indices (remove timezone and set to midnight)
sp500_data.index = pd.to_datetime(sp500_data.index).tz_localize(None).normalize()
btc_data.index = pd.to_datetime(btc_data.index).tz_localize(None).normalize()
nasdaq_data.index = pd.to_datetime(nasdaq_data.index).tz_localize(None).normalize()
tsla_data.index = pd.to_datetime(tsla_data.index).tz_localize(None).normalize()

# Ensure all datasets have the same date range
start_date = max(sp500_data.index.min(), btc_data.index.min(), nasdaq_data.index.min(), tsla_data.index.min())
end_date = min(sp500_data.index.max(), btc_data.index.max(), nasdaq_data.index.max(), tsla_data.index.max())
sp500_data = sp500_data[(sp500_data.index >= start_date) & (sp500_data.index <= end_date)]
btc_data = btc_data[(btc_data.index >= start_date) & (btc_data.index <= end_date)]
nasdaq_data = nasdaq_data[(nasdaq_data.index >= start_date) & (nasdaq_data.index <= end_date)]
tsla_data = tsla_data[(tsla_data.index >= start_date) & (tsla_data.index <= end_date)]

# Rename columns to avoid collision and fit to the desired format
sp500_data = sp500_data.rename(columns={'Close': 'sp500_stockprice', 'Volume': 'sp500_volume'})
btc_data = btc_data.rename(columns={'Close': 'bitcoin_stockprice', 'Volume': 'bitcoin_volume'})
nasdaq_data = nasdaq_data.rename(columns={'Close': 'nasdaq_stockprice', 'Volume': 'nasdaq_volume'})
tsla_data = tsla_data.rename(columns={'Close': 'tesla_stockprice', 'Volume': 'tesla_volume'})

# Plot the adjusted datasets to ensure alignment
plt.figure(figsize=(14, 7))
plt.plot(sp500_data.index, sp500_data['sp500_stockprice'], label='S&P 500 Close', color='blue')
plt.plot(btc_data.index, btc_data['bitcoin_stockprice'], label='Bitcoin Close', color='orange')
plt.plot(nasdaq_data.index, nasdaq_data['nasdaq_stockprice'], label='NASDAQ Close', color='green')
plt.plot(tsla_data.index, tsla_data['tesla_stockprice'], label='Tesla Close', color='purple')
plt.title('Aligned Close Prices Over Time')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.legend()
plt.grid(True)
plt.show()
```

```python
sp500_data.info()
```

```python
# --- Vorbereitung: Date-Spalte sicherstellen ---
for df in [sp500_data, btc_data, nasdaq_data, tsla_data]:
    if 'Date' not in df.columns:
        df.reset_index(inplace=True)  # holt 'Date' aus dem Index
    df['Date'] = pd.to_datetime(df['Date'])

# --- Merge auf stockprice-Preise ---
combined_data_stockprice = sp500_data[['Date', 'sp500_stockprice']].merge(
    btc_data[['Date', 'bitcoin_stockprice']], on='Date', how='outer'
).merge(
    nasdaq_data[['Date', 'nasdaq_stockprice']], on='Date', how='outer'
).merge(
    tsla_data[['Date', 'tesla_stockprice']], on='Date', how='outer'
)

# --- Merge auf Volumen ---
combined_data_volume = sp500_data[['Date', 'sp500_volume']].merge(
    btc_data[['Date', 'bitcoin_volume']], on='Date', how='outer'
).merge(
    nasdaq_data[['Date', 'nasdaq_volume']], on='Date', how='outer'
).merge(
    tsla_data[['Date', 'tesla_volume']], on='Date', how='outer'
)

# --- Nach Datum sortieren (optional, aber empfohlen) ---
combined_data_stockprice.sort_values('Date', inplace=True)
combined_data_volume.sort_values('Date', inplace=True)
```

```python
# Plot the merged stockprice values
plt.figure(figsize=(14, 7))
plt.plot(combined_data_stockprice.index, combined_data_stockprice['sp500_stockprice'], label='S&P 500 stockprice', color='blue')
plt.plot(combined_data_stockprice.index, combined_data_stockprice['bitcoin_stockprice'], label='Bitcoin stockprice', color='orange')
plt.plot(combined_data_stockprice.index, combined_data_stockprice['nasdaq_stockprice'], label='NASDAQ stockprice', color='green')
plt.plot(combined_data_stockprice.index, combined_data_stockprice['tesla_stockprice'], label='Tesla stockprice', color='purple')
plt.title('Merged stockprice values (close prices) Over Time')
plt.xlabel('Date')
plt.ylabel('stockprice value/Close Price')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Merge explizit auf 'Date'
merged_data = pd.merge(
    combined_data_stockprice, combined_data_volume,
    on='Date', how='outer'
)

# Optional: zurücksortieren und Index setzen
merged_data.sort_values('Date', inplace=True)

# Vorschau
print(merged_data.head())
```
